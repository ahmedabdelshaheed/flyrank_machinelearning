## 1. My lane as an ML task (type)

*Classification, clustering, ranking, or scoring — which one, and why?*

Classification + Ranking.

The model first classifies each page by predicting its probability of performance movement. Then, pages are ranked in a prioritized queue based on the model's predicted probability combined with business impact metrics.

## 2. Target or proxy

*What would you predict? Where does that label come from — observed outcome or a defined rule?*


**What would you predict?**
Whether a page will experience a performance decline or growth in the future (`is_declining_label`).

**Where does that label come from — observed outcome or a defined rule?**
It comes from an **observed outcome** calculated directly from performance changes in the `fact_content_daily_performance` table. By comparing a page's metrics (clicks and impressions) in a future evaluation window (e.g., next 30 days) against its historical baseline window (e.g., prior 90 days), we create a binary ground truth label:
- **`1` (Decline):** The page suffers a significant drop (e.g., >20%) in clicks or impressions.
- **`0` (Stable / Growth):** The page maintains or improves its performance.

## 3. Success metric

*One metric you can defend. What number means 'good'?*

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

A fixed rule can flag candidates, but it can't predict whether an update will actually work — success depends on a tangled mix of signals (content type, provider, word count, engagement history) interacting in ways no clean if-statement can capture.
