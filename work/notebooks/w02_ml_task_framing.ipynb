## 1. My lane as an ML task (type)

*Classification, clustering, ranking, or scoring — which one, and why?*

Classification + Ranking.
Each page gets classified into a suggested action (refresh / expansion / protection / pruning / monitoring), then candidate pages are ranked by priority using a model score.

## 2. Target or proxy

*What would you predict? Where does that label come from — observed outcome or a defined rule?*

**What would you predict?**

Whether a page needs an update, what type of update is appropriate (refresh / expansion / protection / pruning / monitoring), and the probability that the update would succeed.

**Where does that label come from — observed outcome or a defined rule?**

Both, for different parts. The "needs update" candidate flag comes from a **defined rule** (position, trend). The "likely to improve" score comes from an **observed outcome**: real before/after performance change for ~26,000 pages that were actually updated in the warehouse's daily performance history.
