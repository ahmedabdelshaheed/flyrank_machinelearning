```python
import pandas as pd, numpy as np
df = pd.read_csv("data/raw/content_refresh_anonymized.csv")

# 1. Trend down — drop nulls in trend_direction only
valid_trend = df.dropna(subset=['trend_direction'])
trend_down = valid_trend[valid_trend['trend_direction'] == 'down']
print(f"Pages with trend_direction = down: {len(trend_down)} out of {len(valid_trend)} ({len(trend_down)/len(valid_trend)*100:.1f}%)")

# 2. Low scroll rate — drop nulls in scroll_rate only
valid_scroll = df.dropna(subset=['scroll_rate'])
low_scroll = valid_scroll[valid_scroll['scroll_rate'] < 30]
print(f"Pages with low scroll rate (<30%): {len(low_scroll)} out of {len(valid_scroll)} ({len(low_scroll)/len(valid_scroll)*100:.1f}%)")

# 3. Target pages — drop nulls in search_volume and avg_position only
valid_target = df.dropna(subset=['search_volume', 'avg_position'])
target_pages = valid_target[(valid_target['avg_position'] > 10) & (valid_target['search_volume'] > 50)]
print(f"Pages with search_volume > 50 and avg_position > 10: {len(target_pages)} out of {len(valid_target)} ({len(target_pages)/len(valid_target)*100:.1f}%)")
```

```python
print(f"Missing trend_direction: {df['trend_direction'].isnull().sum()}")
print(f"Missing scroll_rate: {df['scroll_rate'].isnull().sum()}")
print(f"Missing search_volume: {df['search_volume'].isnull().sum()}")
print(f"Missing avg_position: {df['avg_position'].isnull().sum()}")
```
